```python
from diffusers import StableDiffusionPipeline
import torch

# ============================================================
# LOAD STABLE DIFFUSION MODEL
# ============================================================

model_id = "runwayml/stable-diffusion-v1-5"

pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16,
    use_safetensors=True
)

pipe = pipe.to("cuda")

# ============================================================
# TEXT PROMPT
# ============================================================

prompt = "A futuristic city skyline at sunset, digital art, highly detailed"

# ============================================================
# IMAGE GENERATION
# ============================================================

image = pipe(
    prompt,
    num_inference_steps=30,
    guidance_scale=7.5
).images[0]

# ============================================================
# SAVE AND DISPLAY
# ============================================================

image.save("generated_city.png")

display(image)

print("Image generated successfully!")
print("Saved as: generated_city.png")

# ============================================================
# RESULT
# ============================================================

print("\n========== RESULT ==========")
print("An image generation application using the Stable Diffusion")
print("diffusion model was successfully implemented.")
```
