```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

# ============================================================
# LOAD CODEGEN MODEL
# ============================================================

model_name = "Salesforce/codegen-350M-mono"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

# Use GPU if available
device = "cuda" if torch.cuda.is_available() else "cpu"
model = model.to(device)

print("CodeGen model loaded successfully!")
print("Device:", device)


# ============================================================
# FUNCTION FOR CODE GENERATION
# ============================================================

def generate_code(prompt, max_new_tokens=80):

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to(device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=True,
            temperature=0.7,
            top_p=0.95,
            pad_token_id=tokenizer.eos_token_id
        )

    # Decode generated text
    generated = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return generated


# ============================================================
# 1. CODE GENERATION
# ============================================================

prompt1 = """# Write a Python function to check if a number is prime
def is_prime(n):"""

generated_code = generate_code(
    prompt1,
    max_new_tokens=80
)

print("\n========== CODE GENERATION ==========")
print(generated_code)


# ============================================================
# 2. CODE DEBUGGING
# ============================================================

buggy_code = """# Fix the following Python factorial function.
# The function contains a logical error.

def factorial(n):
    result = 0
    for i in range(1, n + 1):
        result = result * i
    return result

# Corrected function:
def factorial_fixed(n):"""

debugged_code = generate_code(
    buggy_code,
    max_new_tokens=80
)

print("\n========== CODE DEBUGGING ==========")
print(debugged_code)


# ============================================================
# RESULT
# ============================================================

print("\n========== RESULT ==========")
print("An AI-powered code generation and debugging assistant")
print("was successfully implemented using the CodeGen model.")
```
