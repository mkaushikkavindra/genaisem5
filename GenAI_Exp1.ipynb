```python
prompt = "Artificial Intelligence will transform the future of"

outputs = generator(
    prompt,
    max_length=60,
    num_return_sequences=2,
    temperature=0.8,
    top_k=50,
    top_p=0.95,
    do_sample=True
)

print("INPUT:", prompt)
print()

for i, out in enumerate(outputs, 1):
    print(f"--- Generated Text {i} ---")
    print(out["generated_text"])
    print()

print("RESULT:")
print("A text generation application using the pre-trained GPT-2 foundation model")
print("was successfully developed, generating coherent text using sampling-based")
print("decoding strategies.")
```
