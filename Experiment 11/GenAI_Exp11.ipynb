```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from diffusers import StableDiffusionPipeline
from gtts import gTTS
from IPython.display import Audio, display
import torch

# ============================================================
# INPUT TOPIC
# ============================================================

topic = "The benefits of renewable energy"

print("Topic:", topic)


# ============================================================
# 1. TEXT GENERATION USING FLAN-T5
# ============================================================

text_model_name = "google/flan-t5-base"

text_tokenizer = AutoTokenizer.from_pretrained(text_model_name)
text_model = AutoModelForSeq2SeqLM.from_pretrained(text_model_name)

text_prompt = f"Write a short, engaging paragraph about: {topic}"

text_inputs = text_tokenizer(
    text_prompt,
    return_tensors="pt"
)

text_output = text_model.generate(
    **text_inputs,
    max_new_tokens=80
)

generated_text = text_tokenizer.decode(
    text_output[0],
    skip_special_tokens=True
)

print("\n========== GENERATED TEXT ==========")
print(generated_text)


# ============================================================
# 2. IMAGE GENERATION USING STABLE DIFFUSION
# ============================================================

image_model = "runwayml/stable-diffusion-v1-5"

pipe = StableDiffusionPipeline.from_pretrained(
    image_model,
    torch_dtype=torch.float16,
    use_safetensors=True
).to("cuda")

image_prompt = f"An illustration representing {topic}, digital art"

image = pipe(
    image_prompt,
    num_inference_steps=25,
    guidance_scale=7.5
).images[0]

image.save("content_image.png")

print("\n========== GENERATED IMAGE ==========")
print("Image generated and saved as content_image.png")


# ============================================================
# 3. AUDIO GENERATION USING gTTS
# ============================================================

tts = gTTS(
    text=generated_text,
    lang="en"
)

tts.save("content_audio.mp3")

print("\n========== GENERATED AUDIO ==========")
print("Audio generated and saved as content_audio.mp3")


# Optional: play audio in Colab
display(Audio("content_audio.mp3"))


# ============================================================
# RESULT
# ============================================================

print("\n========== RESULT ==========")
print("An integrated AI-based multimedia content generation system")
print("was successfully developed to generate text, image, and audio")
print("from a single topic.")
```
